```python
#checking for outliers
plt.figure(figsize=(6,4))
sns.boxplot(data = df.select_dtypes("number"))
plt.xticks(rotation = 90)
plt.show()
```

```python
#univin analysis
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(df["vin"],kde = True)
plt.title("visual representation of Vin")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.barplot(df["mileage"])
plt.xticks(rotation = 90 )
plt.title("visual representation of mileage")
plt.show()
```

```python

sns.scatterplot( x = "mileage", y = "year", data = df)
plt.title("visual representation of mileage vs year")
plt.show()
```

```python
sns.scatterplot( x = "brand", y = "color", data = df)
plt.title("visual representation of brand vs color")
plt.show()
```

```python
#multi variate analysis
plt.figure(figsize=(10,10))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="Blues",fmt=".2f")
plt.show()
```
